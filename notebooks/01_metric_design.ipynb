1. Metric vs. KPI vs. North Star Metric & OEC
Metric: Any quantifiable, measurable data point you can track (e.g., "button clicks", "daily active users").
KPI (Key Performance Indicator): A high-priority metric that is directly tied to a specific business goal (e.g., "checkout conversion rate"). While all KPIs are metrics, not all metrics are KPIs.
North Star Metric: The single, overarching metric that best captures the core value your product delivers to its customers (e.g., Airbnb's might be "Nights Booked," Spotify's might be "Time Spent Listening").
OEC (Overall Evaluation Criterion): In A/B testing (like in your Project 2), this is the primary metric you use to evaluate if an experiment was successful. The OEC is usually a more immediate, testable metric that rolls up into or directly impacts the North Star metric.
2. Leading vs. Lagging Indicators
Leading Indicator: A predictive metric that signals future outcomes. It tells you what might happen (e.g., "number of items added to a shopping cart" predicts future sales).
Lagging Indicator: A metric that looks back at what has already happened (e.g., total monthly sales).
Revenue: Revenue is a lagging indicator. By the time you measure it, the user behaviors that generated the revenue are already in the past.
3. Vanity Metrics
Definition: A metric that looks impressive on paper but doesn't provide any actionable insight into the actual health or engagement of a product.
Example: "Total registered users" or "App downloads." For instance, an app you use daily might boast "100 million total signups," but if 95% of those accounts are inactive, the number is meaningless. A better, non-vanity metric would be "Daily Active Users."
4. Goodhart's Law
Definition: An adage that states: "When a measure becomes a target, it ceases to be a good measure."
Prediction: It predicts that if you optimize exclusively for a single metric, people will find ways to "game the system" to hit that target, usually at the expense of actual quality or overall business health. (For example, if you make "number of resolved customer support tickets per hour" a strict target, agents might start hastily closing tickets without actually fixing the customers' problems).

Brief Recap:

The North Star (Total Delivered Orders): This is the ultimate goal. For a marketplace, a completed, delivered order is the moment value is actually exchanged between the buyer and the seller.
The Input Metrics (The Levers): The CEO cares about the North Star, but the teams need specific targets they can control:
Marketing drives Demand by trying to increase the Conversion Rate (turning visitors into buyers).
Logistics drives Fulfillment by trying to increase the On-Time Rate (delivering packages when promised).
Customer Experience (CX) drives Quality by trying to increase the Average Review Score.
Seller Ops drives Supply by trying to increase the number of Active Sellers with products listed.
The Guardrails (The Protection): These prevent teams from taking shortcuts.
As hinted at in the text, if Logistics tries to hit a 100% On-Time target by padding the estimate (telling customers it takes 20 days instead of 3), the Promised Delivery Time guardrail will spike, alerting the business that the metric is being "gamed" and the customer experience is actually getting worse.

The vanity-metric filter

1. Total customers since launch (cumulative)

Justification: Cumulative metrics only ever go up and hide current performance issues, offering no actionable insights into present business health.
2. Total GMV since launch (cumulative)

Justification: Like cumulative customers, this is a historical ledger that obscures the actual, ongoing trajectory of the business and cannot inform tomorrow's decisions.
3. Monthly GMV

Justification: This is the primary measure of total value exchanged on a marketplace, directly aligning with business success as long as profitability guardrails are in place.
4. Monthly orders

Justification: It's a useful diagnostic for operational volume, but less robust than GMV because it doesn't account for the actual monetary value of those orders.
5. Monthly new customers

Justification: A crucial input metric for the top of the funnel, but it must be paired with retention and acquisition costs (CAC) to ensure you aren't just pouring water into a leaky bucket.
6. Average review score

Justification: It acts as a vital check (guardrail) against growth tactics that degrade the user experience, ensuring that scale doesn't come at the cost of quality.
7. Number of product categories

Justification: It measures the breadth of your catalog structure but not actual supply or demand, making it highly gameable and disconnected from user value.
8. Monthly active sellers

Justification: A key input for a marketplace, as having an engaged supply side is a prerequisite for generating transactions (GMV).
9. On-time delivery rate

Justification: A critical operational guardrail that ensures logistical quality is maintained, though it must be balanced against aggressive delivery promises so teams don't just "pad" the estimates.
10. Average promised delivery days

Justification: It measures the competitiveness of your logistics offering (an input to conversion), but is useless without the on-time rate holding it accountable.
11. Repeat-purchase rate

Justification: This is one of the strongest indicators of product-market fit; if customers come back, they are finding real, recurring value.
12. Cancellation rate

Justification: A critical negative metric (guardrail) that must be kept low to ensure basic marketplace trust, reliability, and functionality.
